### 1. What is OOP?

#### Object-Oriented Programming is a programming paradigm based on:
- Classes
- Objects
- Encapsulation
- Inheritance
- Polymorphism
- Abstraction

> Python supports all 4.

### 2. Class & Object
- Class → Blueprint/template
- Object → Instance created from class

#### Example

```python
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age
        print("Object created")

s1 = Student("Aditya", 20)
```

### 3. Types of Methods

```python
# Python provides 3 types:
# Instance Method
# Works with object attributes
def info(self):
    print(self.name, self.age)
```

```python
# Class Method
#Works with class variables, not instance
@classmethod
def get_count(cls):
    print(cls.count)
```

```python
# Static Method
# Independent utility
@staticmethod
def greet():
    print("Hello!")
```

> Interview Question:
When should you use static methods?
Ans: When function does not use class or instance data.

### 4. Constructor

```python
# Special method: __init__()
# Called automatically when the object is created.
# Default parameter constructor overloading (Python trick)
class Person:
    def __init__(self, name, age=None, address=None):
        self.name = name
        self.age = age
        self.address = address
```

### 5. Destructor

```python
# Rarely used; Python garbage collector handles memory.
def __del__(self):
    print("Object deleted")
```

### 6. Encapsulation

Binding data + methods inside a class.
Python achieves encapsulation using:

- `_var` (protected)
- `__var` (private — name mangling)

#### Example

```python
class Student:
    def __init__(self):
        self.__marks = 90     # private

    def get_marks(self):
        return self.__marks

    def set_marks(self, m):
        if m >= 0:
            self.__marks = m
```

### 7. Inheritance

Allows child class to use properties of parent class.

#### Types of inheritance in Python

- Single
- Multiple
- Multilevel
- Hierarchical
- Hybrid

#### Example

```python
class A:
    pass

class B(A):    # single inheritance
    pass
```

```python
# Using super()
class Parent:
    def __init__(self):
        print("Parent constructor")

class Child(Parent):
    def __init__(self):
        super().__init__()
        print("Child constructor")
```

### 8. Multiple Inheritance

```python
# Python supports it.
class A: pass
class B: pass
class C(A, B): pass 
```

> Important:
Python uses MRO (Method Resolution Order)
Uses C3 Linearization (interview Q!).

### 9. Polymorphism

#### Method Overriding (runtime polymorphism)

```python
# Child class overrides parent method.
class Animal:
    def sound(self):
        print("Animal sound")

class Dog(Animal):
    def sound(self):
        print("Bark")
```

#### Method Overloading
Python does not support true overloading,

but you can simulate it using:
- Default parameters
- `*args`

```python
def add(self, a, b, c=0):
    return a + b + c
```

### 10. Abstraction

```python
# Hiding implementation using abstract classes.
# Using ABC module
from abc import ABC, abstractmethod

class Shape(ABC):
    @abstractmethod
    def area(self):
        pass
```

### 11. Operator Overloading

Python allows overloading using dunder methods.

#### Examples:

- **Operator Overloading Methods**
  - `+` → `__add__()`
  - `-` → `__sub__()`
  - `*` → `__mul__()`
  - `==` → `__eq__()`
  - `len()` → `__len__()`

#### Example

```python
def __add__(self, other):
    return self.value + other.value
```

### 12. Dunder (Magic) Methods

Special methods starting with `__`.

#### Common ones:

- `__init__` → Constructor (runs when object is created)
- `__str__` → String representation for users (`print(obj)`)
- `__repr__` → Developer/debug representation
- `__len__` → Enables `len(obj)`
- `__eq__` → Enables `==` comparison
- `__add__` → Enables `+` operator
- `__contains__` → Enables `in` keyword
- `__call__` → Makes the object callable like a function
