### 1. What is NumPy?

#### In deep:

**NumPy** (short form of **Numerical Python**) is an open-source Python library used for 
numerical computing. It provides tools for working efficiently with arrays, matrices, and 
mathematical functions. 

NumPy introduces a new data structure called the **ndarray** (n-dimensional array), 
which is much faster and more memory-efficient than Python’s built-in lists for 
numerical operations.

Important Features:

- **Multidimensional Arrays:** 
Efficient storage and manipulation of large datasets (e.g., 1D, 2D, 3D arrays).

- **Mathematical Functions:**
A wide range of mathematical operations — linear algebra, Fourier transforms, 
random number generation, etc.

- **Vectorization:**
Operations on entire arrays without explicit loops, making code cleaner and faster.

- **Interoperability:**
Works well with other libraries like Pandas, Matplotlib, TensorFlow, PyTorch etc.

#### In brief:

NumPy (Numerical Python) is the fastest numerical computing library in Python.
It provides:

- ndarray → fast multi-dimensional arrays
- vectorized operations → no loops, extremely fast
- mathematical functions
- linear algebra
- random number generation

NumPy is written in C, which is why it's much faster than Python lists.

### 2. Why NumPy Instead of Lists?

- **Speed**
  - Lists → Slow  
  - NumPy → **50–100x faster**

- **Memory**
  - Lists → High  
  - NumPy → **Low memory usage**

- **Operations**
  - Lists → Uses loops  
  - NumPy → **Vectorized operations (no loops)**

- **Data Types**
  - Lists → Mixed data types allowed  
  - NumPy → **Homogeneous data types**

- **Machine Learning Support**
  - Lists → Poor support  
  - NumPy → **Essential for ML & scientific computing**

```python
# Performance Comparison 
import numpy as np  
import time

size = 10_000_000 # large data set of 10 million numbers  

# Python Lists  
python_list = list(range(size))  
start = time.time() 

list_squared = [x**2 for x in python_list] # square of all nums 

end = time.time()  
print("Python list time:", end - start, "seconds")  

# NumPy Arrays  
np_array = np.array(python_list)  
start = time.time()  

array_squared = np_array ** 2 # vectorized operation 

end = time.time() 
print("NumPy array time:", end - start, "seconds")
```

```python
# Memory Usage comparison  
import sys
print("Python list size:", sys.getsizeof(python_list) * len(python_list)) 
print("NumPy array size:", np_array.nbytes) 
```

### 6. Data Types in NumPy

We have already discussed how every NumPy array has a **single data type** (homogeneous arrays) & it is stored in the `.dtype` attribute.
Now, let’s look into some of the most **common data types** in NumPy:
- Integer literals : `int32`, `int64` 
- Floating literals : `float32`, `float64`
- Boolean : `bool`
- Complex numbers : `complex64`, `complex128`
- String : `S` (byte-str) & `U` (unicode-str)   
- Object : generic python objects – `object`

```python
# Common Data Types  
arr = np.array([1, 2, 3, 4, 5])  
arr2 = np.array([1.0, 2.0, 3.0])  
arr3 = np.array(["hello", "world", "prime", "ai/ml"])  

print(arr.dtype) # int64 
print(arr2.dtype) # float64  
print(arr3.dtype) # U  

# Complex Numbers  |
arr1 = np.array([2 + 3j])  
arr2 = np.array([5 + 8j])  

print(arr1, arr1.dtype)  
print(arr1 + arr2)  
print(arr2 - arr1)  

# Objects  
arr = np.array (["hello", {1, 2, 3}, 3.14])  
print(arr, arr.dtype) 
```

##### We can also change the data type, either by explicitly typecasting at array creation using `dtype` attribute or by using the `astype` method while creating a new array from an existing one. 

```python
# Changing the data type  
arr = np.array([1, 2, 3, 4, 5])

new_arr = arr.astype("float64")  
print(new_arr, new_arr.dtype)  

new_arr = np.array([1, 2, 3, 4, 5], dtype="float64")  
print(new_arr, new_arr.dtype) 
```

**Why does dtype matter?**
- Memory efficiency 
  - `np.int8` uses 1 byte per element, `np.int64` uses 8 bytes. 
- Performance
  - Smaller types = faster computations. 
- Compatibility 
  - Images often use `np.uint8` 
  - ML libraries expect `float32`

### 8. Broadcasting (VERY IMPORTANT)

**Broadcasting**
Broadcasting allows NumPy to **automatically expand arrays of different shapes** so that
arithmetic operations can be performed. It’s basically scaling arrays without using extra memory. 
- No need to manually reshape arrays. 
- Useful for **combining arrays of different dimensions**.
  
**Broadcasting Rules**
Broadcasting can only take place when the arrays are of compatible shape. So NumPy compares shapes of arrays
from **right to left**. For the array to be compatible, all dimensions must either be:
- Equal, or 
- 1, or 
- Missing (smaller array can be “stretched”).

NumPy applies operations between arrays of different shapes.

##### Example:

```python
arr = np.array([1,2,3])
matrix = np.ones((3,3))

# NumPy expands the smaller array automatically.
matrix + arr
```

### 13. Useful Mathematical Functions
NumPy is massive & provides a **wide range of built-in mathematical functions** that are highly optimized and
can operate **element-wise on arrays**. Let’s have a look at some of them: 

**Aggregation Functions**
These are functions that take an array and reduce it to a single value (or smaller array) 
by combining elements.
- `sum()` - returns sum of all elements
- `prod()` - returns product of all elements 
- `min()` - returns minimum value 
- `max()` - returns maximum value
- `argmin()` - returns index of min value
- `argmax()` - returns index of max value
- `mean()` - returns mean (average)
- `median()` - returns median
- `std()` - returns standard deviation 
- `var()` - returns variance 

```python
arr = np.array([1, 2, 3, 4, 5])

print(np.sum(arr))  # 15 
print(np.prod(arr)) # 120  
print(np.min(arr)) # 1  
print(np.argmin(arr)) # 0 
print(np.max(arr)) # 5  
print(np.argmax(arr)) # 4  
print(np.mean(arr)) # 3.0  
print(np.median(arr)) # 3.0  
print(np.std(arr)) # 1.41  
print(np.var(arr)) # 2.0 
```

**Power Functions**
- `square()` - returns square of all elements
- `sqrt()` - returns square root of all elements
- `pow(a, b)` - returns ab

```python
arr = np.array([1, 2, 3, 4, 5]) 

print(np.square(arr))  # [1, 4, 9, 16, 25] 
print(np.sqrt(arr))  # [1, 1.41, 1.73, 2, 2.23]  
print(np.pow(arr, 3))  # [1, 8, 27, 64, 125]
```

**Log & Exponential Functions**
- `log()` - returns natural log
- `log10()` - returns log base 10 
- `log2()` - returns log base 2
- `exp(x)` - returns ex 

```python
print(np.log(arr))  
print(np.log10(arr))  
print(np.log2(arr))  
print(np.exp(arr)) 
```

**Rounding Functions**
- `round()` - rounds off to nearest value 
- `ceil()` - rounds up 
- `floor()` - rounds down
- `trunc(x)` - truncates(removes) the fractional part  

```python
print(np.round(2.678)) # 3.0  
print(np.floor(2.678)) # 2.0 
print(np.ceil(2.678))  # 3.0  
print(np.trunc(2.678)) # 2.0 
```

**Miscellaneous Functions**
- `abs()` - returns absolute value
- `sort()` - returns sorted(arranges in increasing order) values 
- `unique()` - returns unique values

```python
arr = np.array([1, 2, -5, 3, 8, -4, 2, 5]) 

print(np.abs(arr))  # [1 2 5 3 8 4 2 5]  
print(np.sort(arr)) # [-5 -4 1 2 2 3 5 8] 
print(np.unique(arr)) # [-5 -4 1 2 3 5 8] 
```

### 14. ADVANCED NUMPY

#### 1. Strides

```python
# Strides tell NumPy how many bytes to skip in memory to get to the next element.
arr = np.array([10, 20, 30, 40])
slice_arr = arr[1:4:2]   # take every 2nd element from index 1 → [20, 40]

print("Array:", arr)
print("Slice:", slice_arr)
print("Original Strides:", arr.strides)
print("Slice Strides:", slice_arr.strides)
```

#### 2. Views vs Copies

Slicing a list returns a copy but slicing a NumPy array returns a view - for efficiency. 
View is like a shallow copy that shares the same data as the original array, so no 
duplication happens here. 

- **Views** are fast and memory-efficient (no data duplication). 
- **Copies** are safe but slower and use more memory.

```python
# Slicing gives a VIEW (no new memory)
arr = np.array([1,2,3,4])
view = arr[1:3] # view → points to same memory
view[0] = 99

print(arr) # arr changes!
```

```python
# copy() creates a NEW array
arr = np.array([1,2,3])
copy_arr = arr.copy()
copy_arr[0] = 99

print(arr)       # unchanged
print(copy_arr)
```

#### 7. Performance Tools

##### np.einsum() (Einstein Summation)

```python
# Fast, flexible matrix operations.
a = np.array([1,2,3])
b = np.array([4,5,6])

dot = np.einsum('i,i->', a, b)
print(dot)
```

- `'i'` = index
- `'i,i'` = multiply matching elements
- `'->'` = sum them (because no output indices)
- Result = dot product

##### np.matmul() (matrix multiplication)

```python
A = np.array([[1,2],
              [3,4]])
B = np.array([[5,6],
              [7,8]])

print(np.matmul(A, B))
```

##### numexpr (fast expression evaluation)

```python
# Much faster than regular NumPy operations.
import numexpr as ne
a = np.arange(1000000)
b = np.arange(1000000)

c = ne.evaluate("a * 2 + b * 3")
c
```

##### numba (JIT Compiler — speeds up Python loops)

```python
# Runs 10–100x faster.
from numba import njit

@njit
def fast_sum(arr):
    total = 0
    for x in arr:
        total += x
    return total

print(fast_sum(np.arange(1000000)))
```
