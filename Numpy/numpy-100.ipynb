#### 16. How to add a border (filled with 0's) around an existing array?

````python
arr5 = np.ones((5, 5))
arr5
````

````python
arr5 = np.pad(arr5, pad_width=1, mode='constant', constant_values=0)
arr5
````

#### `pad_width=1`

This means:

- Add 1 layer of padding on every side
- Applies to all dimensions
- For a 2D array:
  - 1 row on top
  - 1 row on bottom
  - 1 column on left
  - 1 column on right

So the padded array becomes larger by +2 in each dimension.

Example:
If `arr5` is shape `(m, n)`, new shape = `(m+2, n+2)`.

#### `mode='constant'`
This means the padding values will be constant (not reflected, repeated, etc.).

#### `constant_values=0`

This sets the padding value to 0.
So the border is filled with zeros.

#### 17. What is the result of the following expression? (★☆☆)
```python
0 * np.nan
np.nan == np.nan
np.inf > np.nan
np.nan - np.nan
np.nan in set([np.nan])
0.3 == 3 * 0.1
```

````python
# Any arithmetic operation involving NaN produces NaN
print(0 * np.nan)
````

````python
# NaN is not equal to anything, including itself.
print(np.nan == np.nan)
````

````python
# Any comparison involving NaN returns False, except !=, which returns True.
# np.nan is unordered: it is not greater, less, or equal to anything.
print(np.inf > np.nan)
````

````python
# Any arithmetic with NaN → NaN.
print(np.nan - np.nan)
````

````python
print(np.nan in set([np.nan]))
````

Output: `True`, surprisingly!

Why:
This seems contradictory because `np.nan` != `np.nan`, but:

#### Sets use hashing for membership
- `np.nan` has a stable hash value.
- The hash is used first to find the bucket in the set.
- Equality is checked only afterward—but since it's the same object, it is treated as found.

So `x in set([x])` is always true, even if `x != x`.

> Using NaN as a set key works, but using multiple NaNs can behave oddly.

````python
print(0.3 == 3 * 0.1)
````

Output: `False`

Why:
This is due to binary floating-point precision.

- `0.1` cannot be represented exactly in binary.
- `3 * 0.1` becomes `0.30000000000000004`

This is not exactly equal to 0.3

````python
print(3 * 0.1)
````

#### 27. Consider an integer vector Z, which of these expressions are legal?
```python
Z**Z
2 << Z >> 2
Z <- Z
1j*Z
Z/1/1
Z<Z>Z
```

- **`Z**Z`** → Legal  
  - Performs element-wise exponentiation

- **`2 << Z >> 2`** → Legal  
  - Applies element-wise bitwise left/right shifts

- **`Z <- Z`** → Illegal  
  - `<-` is not a valid operator in Python

- **`1j * Z`** → Legal  
  - Complex scalar multiplied with array, result becomes complex

- **`Z/1/1`** → Legal  
  - Standard element-wise division

- **`Z < Z > Z`** → Illegal  
  - NumPy arrays cannot be used in chained comparisons

#### 29. How to round away from zero a float array ?

````python
arr10 = np.random.uniform(-10,+10,10)
print(np.copysign(np.ceil(np.abs(arr10)), arr10))

# More readable but less efficient
print(np.where(arr10>0, np.ceil(arr10), np.floor(arr10)))
````

- **`np.copysign(np.ceil(np.abs(arr10)), arr10)`**
  - Uses a math trick: take absolute value → apply `ceil()` → restore original sign
  - Produces the same result as the conditional method
  - Very fast (fully vectorized)

- **`np.where(arr10 > 0, np.ceil(arr10), np.floor(arr10))`**
  - Uses if–else branching: positive → `ceil()`, negative → `floor()`
  - Produces the same result
  - Slower due to conditional checks

---

- **`np.ceil(x)`**
  - Rounds *up* to the next integer
  - Examples:
    - `np.ceil(3.2)` → `4`
    - `np.ceil(-3.2)` → `-3`
- **`np.floor(x)`**
  - Rounds *down* to the previous integer
  - Examples:
    - `np.floor(3.7)` → `3`
    - `np.floor(-3.7)` → `-4`
- **`np.round(x)`**
  - Rounds to the nearest integer (uses banker’s rounding)
  - Examples:
    - `np.round(3.5)` → `4`
    - `np.round(2.5)` → `2`
- **`np.trunc(x)`**
  - Truncates decimal part → rounds *toward zero*
  - Examples:
    - `np.trunc(3.8)` → `3`
    - `np.trunc(-3.8)` → `-3`

#### 34. Consider a generator function that generates 10 integers and use it to build an array

````python
def generate():
    for x in range(10):
        yield x
arr11 = np.fromiter(generate(),dtype=float,count=-1)
arr11
````

- `generate()` yields numbers 0–9.
- `np.fromiter()` consumes the generator and creates a NumPy array.
`count=-1` → read all items.
- Output is a float array
