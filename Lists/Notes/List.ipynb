## Python Lists – Complete Guide

### A List in Python is:

- A list is a collection of items in Python. 
- Items in a list are ordered, changeable (mutable), and can contain duplicates. 
- Lists can hold any data type: numbers, strings, other lists, etc. 
- Lists are written using square brackets `[]`

Example:

````python
my_list = [10, "hello", 3.14, True]
my_list
````

### List Characteristics 
- **Ordered** – Items have a defined order and can be accessed by index. 
- **Mutable** – Items can be changed after the list is created. 
- **Allows duplicates** – Same value can appear more than once. 
- **Heterogeneous** – Can contain different data types.

### 2. Accessing List Items

#### List Indexing 
**Index** in list is the position value of an item. Index starts from `0`.  
We can use index to access elements, modify the list or even slice it.

````python
# Indexing (0-based)
a = [10, 20, 30]
print(a[0])
print(a[-1])
````

#### Slicing
**Slicing** in lists is same as slicing in strings.  
The general syntax for slicing a list is:

`list[start:end:step]`
- `start` - inclusive 
- `end` - exclusive 
- `step` - optional (default = 1)

````python
# Slicing
a = [10, 20, 30, 40, 50]
print(a[1:4])
print(a[:3])
print(a[3:])
print(a[::2])  
````

### 9. List Comprehension

````python
# Basic
squares = [x*x for x in range(1, 6)]
squares
````

````python
# With condition
evens = [x for x in range(20) if x % 2 == 0]
evens
````

````python
# If-else inside list comprehension
result = ["even" if x%2==0 else "odd" for x in range(5)]
result
````

````python
# Nested list comprehension
pairs = [(i,j) for i in range(2) for j in range(3)]
pairs
````

### List Methods (Complete Table)

- **`append(x)`** – Add element at the end  
- **`extend(iterable)`** – Add multiple elements  
- **`insert(i, x)`** – Insert element at a specific index  
- **`remove(x)`** – Remove the first occurrence of `x`  
- **`pop(i)`** – Remove and return element at index `i`  
- **`clear()`** – Remove all items from the list  
- **`index(x)`** – Return index of element `x`  
- **`count(x)`** – Count occurrences of an element  
- **`sort()`** – Sort the list  
- **`reverse()`** – Reverse the list  
- **`copy()`** – Create a shallow copy of the list

### Using lists as queues (Not efficient)
A queue follows FIFO → First In, First Out

Meaning:
- Add elements at end
- Remove elements from the front

#### Using list:
Adding (append):
```python
queue.append(x)
```

- Fast
- O(1)

#### Removing from front:
```python
queue.pop(0)
```

- Very slow
- O(n) time
- Shifts all elements one step left

#### Why pop(0) is Bad (Slow)

In memory, a list is stored like this:

```css
[ a | b | c | d | e ]
```

If you remove index 0, Python must shift all elements:

```css
[ b | c | d | e ]
```

This shifting makes it O(n) each time you pop from the front.

Queues need to be fast → but using lists makes them slow.

#### Better: use collections.deque.
