```python
# import libraries
import requests
from bs4 import BeautifulSoup
```

```python
# fetch the data and store it
page_count = 1

while True:
    URL = f"https://quotes.toscrape.com/page/{page_count}/"
    res = requests.get(URL)

    soup = BeautifulSoup(res.text, "lxml")
    quotes = soup.select("div.quote")

    if not quotes:
        print("No valid pages anymore....")
        break

    with open(f"Quotes/quotes{page_count}.html", "w", encoding="utf-8") as f:
        f.write(res.text)
        print(f"Downloaded data from page{page_count}")

    page_count = page_count + 1
```

```python
# Extract data
life_quotes = []
page_count = 1

while page_count <= 10:
    with open(f"Quotes/quotes{page_count}.html", "r", encoding="utf-8") as f:
        html_content = f.read()
    
    soup = BeautifulSoup(html_content, "lxml")
    all_quotes = soup.select("div.quote")
    
    for q in all_quotes:
        # Get all tags for this specific quote
        all_tags = [tag.get_text() for tag in q.select(".tags .tag")]
        
        # Check the condition AFTER gathering all tags for the quote
        if "life" in all_tags:
            text = q.select_one("span.text").get_text()
            author = q.select_one("small.author").get_text()
            life_quotes.append([text, author])
            
    page_count += 1 # Increment at the end of the loop
```

```python
for quote in life_quotes:
    print(quote)
```
