## Collecting Data Using Web Scraping for Data Science

### 1. What is Web Scraping?

Web scraping is the process of automatically extracting data from websites using code.

**In simple terms:**

> Browser shows data → Scraper extracts data → You analyze it

### 2. Why Web Scraping is Used in Data Science

**Web scraping is used when:**
- No API is available
- Data is publicly visible but not downloadable
- You need custom or niche datasets

**Common Use Cases**
- Job market analysis
- Product price tracking
- News & sentiment analysis
- Sports statistics
- Real estate data

### 3. Web Scraping vs APIs
- **Data access**
  - Web Scraping: HTML pages
  - API: Structured endpoints

- **Speed**
  - Web Scraping: Slower
  - API: Faster

- **Stability**
  - Web Scraping: Can break
  - API: More stable

- **Ethics**
  - Web Scraping: Must be careful
  - API: Usually permitted

> Prefer APIs first, scrape only if needed

### 4. How Web Scraping Works (Flow)
`Request → HTML → Parse → Extract → Store → Analyze`

### 5. Tools & Libraries for Web Scraping

**Python Libraries**
- `requests` – Fetch web pages
- `BeautifulSoup` – Parse HTML
- `lxml` – Fast HTML parsing
- `Selenium` – Dynamic websites (JS)
- `Scrapy` – Large-scale scraping

> Most data scientists use requests + BeautifulSoup

### 6. Basic Web Scraping Example

**Install Libraries**
> !pip install requests beautifulsoup4

````python
# Step 1: Fetch Web Page
import requests

url = "https://www.scrapethissite.com/pages/simple/"
response = requests.get(url)
html = response.text
````

````python
# Step 2: Parse HTML
from bs4 import BeautifulSoup
soup = BeautifulSoup(html, "html.parser")
````

````python
# Step 3: Extract Data
all_h3 = soup.find_all("h3")

all_countries = []

for h3 in all_h3:
    name = h3.get_text(strip=True)
    # print(h3.find_parent("div")["class"])

    population = h3.find_next("div").select("span.country-population")[0].get_text(strip=True)
    # print(h3.find_next("div").select_one("span.country-population").get_text(strip=True))

    all_countries.append([name, population])
````

````python
# Step 4: convert it into dataframe
import pandas as pd
df = pd.DataFrame(all_countries, columns=["Name", "Population"])
df.head()
````

````python
# save data in csv format
df.to_csv("scrap.csv", index=False)
````

### 7. Understanding HTML Structure (Very Important)

**Common Tags**
- `<div>` – Containers
- `<span>` – Inline text
- `<a>` – Links
- `<table>` – Tabular data

**Selecting Elements**
```python
soup.find("div", class_="price")
soup.find_all("a")
```

### 8. CSS Selectors (Cleaner Way)
```python
soup.select("div.price")
soup.select("table tr")
```
> CSS selectors are powerful and preferred

### 9. Scraping Tables (Very Common)

### 11. Scraping Dynamic Websites (JavaScript)

### 12. Dealing with Blocks & Anti-Scraping

**Common Protections**
- IP blocking
- CAPTCHA
- Rate limits

**Best Practices**
- Add delays
- Rotate user agents
- Respect robots.txt

```python

time.sleep(2)
```

### 13. Ethics & Legal Considerations (VERY IMPORTANT)

- Scrape only public data
- Read Terms of Service
- Respect robots.txt
- Do not overload servers
- Do not scrape private data

> Ethical scraping is mandatory
