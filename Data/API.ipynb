## Collecting Data Using APIs for Data Science

### 1. What is an API?

An **API (Application Programming Interface)** allows two software systems to communicate with each other.

In data science, APIs are used to:

- Fetch real-time data
- Access large datasets programmatically
- Automate data collection pipelines

**Simple idea:**
> You send a request → API returns data (usually JSON)

### 2. Why APIs Are Important in Data Science

**APIs are used when:**
- Data is dynamic (weather, stock prices)
- Data is too large for manual download
- Data needs regular updates

**Common Use Cases**
- Financial data (stocks, crypto)
- Social media analytics
- Weather prediction
- Machine learning datasets
- Web & mobile analytics

### 3. Types of APIs

**REST API (Most Common)**
- Uses HTTP methods
- Stateless
- Returns JSON/XML

**SOAP API**
- XML-based
- More rigid
- Less used in data science

**GraphQL API**
- Fetch exactly what you need
- Efficient for large datasets

> Data science mostly uses REST APIs

### 4. HTTP Methods Used in APIs

- **GET** – Retrieve data  
- **POST** – Send data  
- **PUT** – Update data  
- **DELETE** – Remove data
> For data collection → mostly GET requests

### 5. API Request Structure
```
https://api.example.com/v1/resource?param=value
```

**Components**
- Base URL – API endpoint
- Path – Specific resource
- Query Parameters – Filters
- Headers – Authentication & metadata

### 6. Authentication Methods

Most APIs require authentication to control access.

**Common Methods**

- API Key
```
headers = {
  "Authorization": "Bearer YOUR_API_KEY"
}
```

- OAuth 2.0
  - Used by Google, Twitter, GitHub
  - Token-based

- Basic Auth
   - Username + password

> Always keep keys private

### 7. Data Formats Returned by APIs

- JSON (Most Common)
   - Lightweight
   - Easy to parse

- XML
   - Heavier
   - Structured

- CSV
   - Ready for analysis

> Data science prefers JSON

### 8. Using APIs in Python (Step-by-Step)

````python
# Required Libraries
import requests
import pandas as pd
````

````python
# Step 1: Make an API Request
url = "https://stephen-king-api.onrender.com/api/books"
response = requests.get(url)
````

````python
# Step 2: Check Response Status
response.status_code
````

- **200** – Success  
- **401** – Unauthorized  
- **404** – Not Found  
- **429** – Rate limit exceeded  

````python
# Step 3: Parse JSON Data
data = response.json()
````

````python
# Step 4: Convert to DataFrame
df = pd.DataFrame(data)
df.head()
````

### 9. Handling Query Parameters

The `params` argument is used to send query parameters in a GET request to filter or customize the data returned by an API.
> You’re telling the API:
“Give me data only between 1 Jan 2024 and 31 Dec 2024.”

```python
params = {
    "start_date": "2024-01-01",
    "end_date": "2024-12-31"
}

response = requests.get(url, params=params)
```

### 10. Pagination (Very Important)

APIs often limit records per request.

**Example:**
```python
page = 1
all_data = []

while True:
    response = requests.get(url, params={"page": page})
    data = response.json()
    if not data:
        break
    all_data.extend(data)
    page += 1
```

### 11. Rate Limiting

APIs restrict requests per minute/hour.

**How to Handle**
- Add delays
- Monitor headers
- Retry logic

```python
import time
time.sleep(1)
```

### 12. Error Handling

````python
if response.status_code != 200:
    print("Error occurred")
````

````python
# Better approach:
response.raise_for_status()
````
