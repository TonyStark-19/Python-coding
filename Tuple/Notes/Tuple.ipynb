## PYTHON TUPLES –  Complete Guide

##### What is a Tuple?

A tuple is an ordered, immutable collection of items.

##### Example:

```python
t = (1, 2, 3)
type(t)
```

### 1. Key Features

- **Ordered** – Elements maintain their index positions  
- **Immutable** – Cannot modify, add, or delete elements  
- **Allows duplicates** – Example: `(1, 2, 1)`  
- **Any data types allowed** – Example: `(1, "hello", 3.5)`  
- **Faster than lists** – Due to immutability and internal optimizations  

### 2. Creating Tuples

```python
# Regular tuple
t = (1, 2, 3)
t
```

```python
# Tuple without parentheses (tuple packing)
t = 1, 2, 3
t
```

```python
# Empty tuple
t = ()
t
```

```python
# Single-element tuple
# (You MUST use a comma)
t = (5,)
t
```

### 3. Accessing Items

```python
# Indexing
t = (10, 20, 30)
print(t[1]) 
```

```python
# Slicing
print(t[1:])
```

### 4. Tuple Methods

```python
# Tuples have very few methods because they are immutable.
t = (1,2,3,1)
print(t.count(1))
print(t.index(3))
```

### 5. Tuple Unpacking

```python
# Normal unpacking
a, b, c = (1, 2, 3)
print(a, b, c)
```

```python
# Star unpacking
a, *b = (1, 2, 3, 4)
print(a, b)
```

### 6. Why Tuples Are Useful

#### Immutability

You cannot change tuple values, which makes tuples safe for:
- Fixed data
- Dictionary keys
- Data integrity

#### Memory Efficient (lighter than lists)

- Tuples use less memory.

#### Faster

- Operations on tuples are faster because Python knows they won’t change.

### 7. List vs Tuple

#### **List**
- Mutable  
- Slower  
- More built-in methods  
- Uses more memory  

#### **Tuple**
- Immutable  
- Faster  
- Very few built-in methods  
- Uses less memory  

### 8. Tuples Inside Lists / Lists Inside Tuples

```python
# List inside a tuple (allowed because a list is mutable)
t = (1, [2, 3], 4)
t[1].append(5)
t
```

```python
# Tuple inside list
lst = [(1,2), (3,4)]
lst
```

### 9. Tuple Concatenation

```python
t1 = (1, 2)
t2 = (3, 4)
print(t1 + t2)
```

### 10. Tuple of Tuples

```python
matrix = ((1,2,3), (4,5,6))
matrix
```

### 11. When to Use Tuples

#### Use tuples when:

- Data should not change
- Performance matters
- You want to use data as dictionary keys
- Returning multiple values from a function

#### Example:

```python
def divide(a, b):
    return a//b, a%b

#above function returns a tuple of (q, r)
#which is unpacked here
q, r = divide(10, 3)
print(q, r)
```
