## 1. What is a Dictionary?

A dictionary is a collection of key–value pairs.

- Keys are unique
- Values can be anything (int, list, tuple, another dict…)
- Keys must be immutable (str, int, tuple)

#### Example:

```python
student = {
    "name": "Aditya",
    "age": 20,
    "skills": ["Python", "JS"]
}
type(student)
```

### 2. Creating Dictionaries

```python
# Using {}
d = {"a": 1, "b": 2}
d
```

```python
# Using dict()
d = dict(a=1, b=2)
d
```

```python
# Empty dictionary
d = {}
d
```

```python
# OR
d = dict()
d
```

### 3. Accessing Values

```python
# Using key:
student["name"]
```

```python
# If key doesn’t exist → KeyError
# Using get() (safe)
student.get("age")
```

```python
# None (no error)
student.get("city")
```

```python
# returns default
student.get("city", "Not found")
```

### 4. Adding & Updating Items

```python
# Add new key-value pair:
student["city"] = "Delhi"
student
```

```python
# Update existing key:
student["age"] = 21
student
```

```python
# Update multiple keys:
student.update({"course": "B.Tech", "year": 2})
student
```

### 5. Removing Items

```python
# pop(key) → removes and returns value
# Error if the key does not exist
student.pop("age")
```

```python
# popitem() → removes last inserted item
student.popitem()
```

```python
# del → removes key
del student["skills"]
student
```

```python
# clear() → empties dictionary
student.clear()
student
```

### 6. Dictionary Methods

- **`keys()`** – returns all keys  
- **`values()`** – returns all values  
- **`items()`** – returns key-value pairs  
- **`get(key)`** – safe access without raising an error  
- **`update()`** – update multiple key-value pairs  
- **`pop(key)`** – remove a key and return its value  
- **`popitem()`** – remove the last inserted key-value pair  
- **`clear()`** – empties the dictionary

##### Example:

```python
student = {
    "name": "Aditya",
    "age": 20,
    "skills": ["Python", "JS"]
}
student.keys()  
```

```python
student.values()
```

```python
student.items()
```

### 7. Looping Through Dictionary

```python
# Loop through keys:
for key in student:
    print(key)
```

```python
# Loop through values:
for val in student.values():
    print(val)
```

```python
# Loop through key–value pairs:
for k, v in student.items():
    print(k, v)
```

### 8. Useful Real Examples

```python
# Count frequency of characters
s = "hello"
freq = {}

for ch in s:
    freq[ch] = freq.get(ch, 0) + 1

print(freq)
```
