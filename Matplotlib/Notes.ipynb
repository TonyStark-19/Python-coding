## Matplotlib – Complete Guide for Data Science
Matplotlib is the **foundational plotting library in Python** used for data visualization in data science, machine learning, and research.

### 1. What is Matplotlib?

Matplotlib is a Python library used to create static, animated, and interactive visualizations.

**It works closely with:**
- NumPy
- Pandas
- SciPy

> Think of Matplotlib as the base layer of all Python plotting libraries.

### 2. Why Matplotlib is Important in Data Science

- Understand data visually (EDA)
- Detect patterns & trends
- Identify outliers
- Communicate insights clearly

> Almost every data science project uses Matplotlib.

### 4. Importing Matplotlib
> `pyplot` provides a MATLAB-like interface.

````python
import matplotlib.pyplot as plt
````

### 5. Basic Plot Structure
```python
plt.plot(x, y)
plt.show()
```
- `plot()` → draw
- `show()` → display

### 6. Line Plot
> Used for trends over time

````python
x = [1, 2, 3, 4]
y = [10, 20, 25, 30]

plt.plot(x, y)
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.title("Line Plot Example")
plt.show()
````

### 7. Scatter Plot
> Used for relationship between variables

````python
plt.scatter(x, y)
plt.title("Scatter Plot")
plt.show()
````

### 8. Bar Chart
> Used for categorical comparison

````python
categories = ['A', 'B', 'C']
values = [10, 15, 7]

plt.bar(categories, values)
plt.show()
````

### 9. Histogram
> Used to see data distribution

````python
plt.hist(values, bins=5)
plt.show()
````

### 10. Pie Chart
> Use sparingly in data science

````python
plt.pie(values, labels=categories, autopct='%1.1f%%')
plt.show()
````

### 11. Figure & Axes (Object-Oriented Approach)
> Recommended for complex plots

````python
fig, ax = plt.subplots()
ax.plot(x, y)
ax.set_title("OO Plot")
plt.show()
````

### 12. Customizing Plots

````python
plt.plot(x, y, linestyle='--', marker='o')
plt.grid(True)
plt.show()
````

**Customization options:**
- Line style
- Markers
- Grid
- Labels

### 13. Multiple Plots

````python
plt.plot(x, y)
plt.plot(x, [v*2 for v in y])
plt.show()
````

### 14. Subplots

````python
plt.subplot(1, 2, 1)
plt.plot(x, y)

plt.subplot(1, 2, 2)
plt.scatter(x, y)
plt.show()
````

### 15. Working with Pandas
> Pandas uses Matplotlib internally
```python
df['sales'].plot(kind='line')

```

### 16. Annotations
> Highlight important points
```python
plt.annotate("Peak", xy=(3,25), xytext=(2,30), arrowprops=dict())
plt.show()
```

