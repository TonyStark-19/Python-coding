## Seaborn – Complete Guide
Seaborn is a high-level Python data visualization library **built on top of Matplotlib.** 
It provides a simpler and more attractive way to create statistical graphics, with default 
themes and color palettes that look modern and clean.

**Key Features**
- **Integration with Pandas** – Works seamlessly with DataFrames and handles column 
names directly. 
- **Complex plots made easy** – Automatically handles things like regression lines, 
aggregation, and confidence intervals. 
- **Statistical plotting** – Supports visualization of distributions, relationships, and 
categorical data (e.g., `boxplot`, `violinplot`, `heatmap`)
- Beautiful default styles & in-built datasets for learning.

### 1. Why Seaborn?
- Built-in themes & color palettes
- Works directly with Pandas DataFrames
- Excellent for statistical visualization
- Less code than Matplotlib
- Handles aggregation automatically

### 7. Relational Plots
> Relational plots are plots that show the relationship between two (or more) variables, 
typically numerical x and y variables.
#### Scatter Plot

````python
sns.scatterplot(x="total_bill", y="tip", data=df)
plt.show()
````

````python
# With hue & style:
sns.scatterplot(
    x="total_bill", y="tip",
    hue="sex", style="time",
    data=df
)
plt.show()
````

#### Line Plot

````python
# Automatically shows confidence intervals.
sns.lineplot(x="day", y="total_bill", data=df)
plt.show()
````

#### relplot (Figure-level)

````python
sns.relplot(
    x="total_bill", y="tip",
    hue="sex",
    col="time",
    kind="scatter",
    data=df
)
plt.show()
````

### 8. Categorical Plots
> Categorical plots are used to visualize data where one variable is categorical (discrete) 
and the other is usually numerical. 
#### Bar Plot

````python
# Shows mean + confidence interval
sns.barplot(x="day", y="total_bill", data=df)
plt.show()
````

#### Count Plot

````python
sns.countplot(x="day", data=df)
plt.show()
````

#### Box Plot
**Shows:**
- `Median`
- `Quartiles`
- `Outliers`

````python
sns.boxplot(x="day", y="total_bill", data=df)
plt.show()
````

#### Violin Plot

````python
# Shows distribution shape.
sns.violinplot(x="day", y="total_bill", data=df)
plt.show()
````

#### Strip Plot

````python
sns.stripplot(x="day", y="total_bill", data=df)
plt.show()
````

#### Swarm Plot

````python
# (No overlapping points)
sns.swarmplot(x="day", y="total_bill", data=df)
plt.show()
````

#### catplot (Figure-level)

````python
sns.catplot(
    x="day", y="total_bill",
    kind="box",
    data=df
)
plt.show()
````

### 9. Distribution Plots
> Distribution plots are used to visualize the distribution of a single numerical variable.
#### Histogram

````python
sns.histplot(df["total_bill"], bins=20, kde=True)
plt.show()
````

#### KDE Plot

````python
sns.kdeplot(df["total_bill"], fill=True)
plt.show()
````

#### Displot (Figure-level)

````python
sns.displot(df["total_bill"], kde=True)
plt.show()
````

### 10. Regression & Trend Plots

#### lmplot

````python
sns.lmplot(
    x="total_bill", y="tip",
    hue="sex",
    data=df
)
plt.show()
````

#### regplot

````python
sns.regplot(x="total_bill", y="tip", data=df)
plt.show()
````

### 11. Matrix & Correlation Plots
> Matrix plots are used to visualize 2D data arranged in a matrix or table, such as 
correlation matrices, confusion matrices, or pivot tables.
#### Heatmap

````python
corr = df.corr(numeric_only=True)
sns.heatmap(corr, annot=True, cmap="coolwarm")
plt.show()
````

#### Clustermap

````python
sns.clustermap(corr)
plt.show()
````

### 12. Multi-Plot Grids

#### Pair Plot

````python
# Great for EDA.
sns.pairplot(df, hue="sex")
plt.show()
````

#### FacetGrid

````python
g = sns.FacetGrid(df, col="time", row="sex")
g.map(sns.scatterplot, "total_bill", "tip")
plt.show()
````

### 13. Working with Pandas
Seaborn works directly with DataFrames:

````python
# No manual slicing needed.
sns.boxplot(data=df, x="day", y="tip")
plt.show()
````

### 15. When to Use Seaborn vs Matplotlib
- **Seaborn**
  - Statistical plots
  - Quick EDA
  - Beautiful defaults
  - Works directly with DataFrames

- **Matplotlib**
  - Low-level control
  - Custom visuals
  - High performance
  - Supports complex layouts

> Best practice:
Use Seaborn + Matplotlib together

### 16. Common Mistakes
- Forgetting `data=df`
- Using Seaborn for heavy custom animations
- Mixing figure-level & axes-level incorrectly
- Overplotting without `hue` or `alpha`

### 17. Seaborn vs Plotly
- **Seaborn**
  - Static plots
  - Faster EDA
  - Lightweight
  - Publication-ready visuals

- **Plotly**
  - Interactive plots
  - Dashboards
  - Heavy (more dependencies)
  - Web app–friendly visuals
