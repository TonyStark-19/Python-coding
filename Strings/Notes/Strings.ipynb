## Python String – Complete Guide

### 1. What is a String? 
A string in Python is a sequence of characters enclosed in quotes:
Strings are **immutable**, meaning once created, their contents can’t be changed directly.

```python
s = "Hello Python"
type(s)
```

### 2. How to Create Strings

```python
s1 = "hello"
s2 = 'hello'
s3 = """multi
line
string"""
print(s1)
print(s2)
print(s3)
```

### 3. String Indexing
Index in a sequence (like strings) represents the position value of individual elements 
i.e. characters in case of a string. So Indexing lets us access individual characters in 
a string.
Python follows **Zero-based indexing** i.e. first character is at index `0`. 

```python
s = "Python"
s[0]
```

```python
s[-1]
```

### 4. String Slicing
Slicing is a powerful feature of Python that lets us access multiple elements at once. 
We can do slicing in strings & even on other sequences like lists & tuples. 
The general syntax for slicing a string is:
`string[start : stop : step]`

Where: 
- **start** - index where the slice starts (inclusive). Defaults to `0` if omitted.  
- **stop** - index where the slice ends (exclusive). Defaults to the end of the string if omitted. 
- **step** - how many indices to move forward each time. Defaults to `1`.

**Example:**

```python
s = "Python"
s[1:4]
```

```python
s[:3]
```

```python
s[3:]
```

```python
# reverse string
s[::-1]
```

### 6. Important String Methods

#### Case Methods
- `.upper()` → Convert to uppercase  
- `.lower()` → Convert to lowercase  
- `.title()` → Capitalize each word  
- `.capitalize()` → Capitalize first letter  
- `.swapcase()` → Swap lowercase ↔ uppercase  

#### Search Methods
- `.find(sub)` → Returns first index OR -1  
- `.rfind(sub)` → Returns last index  
- `.index(sub)` → Like find but raises error if not found  
- `.count(sub)` → Counts occurrences of a substring  

#### Check Methods (Boolean)
- `.isalpha()` → Only letters  
- `.isdigit()` → Only digits  
- `.isalnum()` → Letters + digits  
- `.islower()` → All lowercase  
- `.isupper()` → All uppercase  
- `.isspace()` → Only spaces  
- `.istitle()` → Title case  

#### Modification Methods
- `.replace(a, b)` → Replace substring  
- `.strip()` → Remove spaces (left + right)  
- `.lstrip()` → Remove left-side spaces  
- `.rstrip()` → Remove right-side spaces  

#### Split and Join

```python
"hello world".split()
```

```python
"1,2,3".split(",")
```

```python
"-".join(["a","b","c"])
```

### 7. Formatting Strings
String formatting is the process of creating `dynamic` strings by inserting values from 
variables or expressions into a predefined string template. This allows for the 
construction of flexible and informative output based on changing data. 
We have multiple ways to format a string, the most modern 2 are:
- using the `format()` function.
- using `f-strings`

#### Using f-strings (Python 3.6+) 
F-strings are concise, readable & will be our preferred way going forward. 
In f-strings we prefix the string with `f` and put our variable or expressions inside `{}`

```python
# f-strings (best!)
name = "Aditya"
age = 20
print(f"My name is {name} and I am {age}")
```

```python
# You can put any valid Python expression inside {}:
a = 5  
b = 10  
print(f"sum of {a} & {b} = {a + b}")  
print(f"avg of {a} & {b} = {(a + b) / 2}")
```

#### Using `.format()`
In this way we use `{}` as placeholders & pass placeholder replacement values in the 
format function. 

```python
# format()
"{} scored {}".format("Aditya", 95)
```

```python
# We can also use positional & named placeholders: 
"Coordinates: {1}, {0}".format("x", "y")
```

```python
"Name: {n}, Age: {a}".format(n="Bob", a=30) 
```

```python
# % formatting (old)
"%s is %d years old" % ("Aditya", 20)
```

### 16. Common String Problems

```python
# Count character frequency
s = "aditya"
freq = {}
for c in s:
    freq[c] = freq.get(c, 0) + 1

freq
```

```python
# Check palindrome
s == s[::-1]
```

```python
# Find all occurrences of a substring
s = "abababa"
i = 0
while True:
    i = s.find("ab", i)
    if i == -1:
        break
    print(i)
    i += 1
```
