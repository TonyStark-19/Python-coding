## SETS IN PYTHON — Complete Guide

#### What is a Set?

A set is an unordered collection of unique elements in Python.

#### Example:

```python
s = {1, 2, 3}
type(s)
```

### 1. Key Features of Sets

- Unordered → No index, no position
- Unique elements only → Duplicates are removed
- Mutable → You can add or remove items (but elements inside must be immutable)
- Unindexed → You cannot use indexing like `s[0]`
- Fast for membership tests `(in)` because sets use hashing
- Cannot contain mutable items
   - Valid: numbers, strings, tuples
   - Invalid: lists, dictionaries, sets

### 2. Creating Sets

```python
# Using curly braces
s = {1, 2, 3}
s
```

```python
# Using set() constructor
s = set([1, 2, 3])
s
```

```python
# Empty set
s1 = {}       # NOT a set → this creates a dictionary
s2 = set()    # correct
type(s1)
```

```python
type(s2)
```

### 3. Set Methods

```python
# Adding Elements
s3 = set()
x = 10
s3.add(x)        # Adds x
s3.update([1,2]) # Adds multiple values
s3
```

```python
# Removing Elements
# Removes x, error if not found
s3.remove(x)
s3
```

```python
# Removes x, no error if missing
s3.discard(x)
s3
```

```python
# Removes random element
s3.pop()
s3
```

```python
# Empties the set
s3.clear()
s3
```

### 4. Searching

```python
# True/False
s3 = set()
x = 10
s3.add(x)
s3.update([1,2])
x in s3
```

### 5. Set Operations

```python
a = set([1, 2, 3])
b = set([3, 4, 5])
print(a)
print(b)
```

```python
# Union (combine)
a | b
```

```python
a.union(b)
```

```python
# Intersection (common values)
a & b
```

```python
a.intersection(b)
```

```python
# Difference (elements in a but not in b)
a - b
```

```python
a.difference(b)
```

```python
# Symmetric Difference (values not common)
a ^ b
```

```python
a.symmetric_difference(b)
```

### 6. Other Useful Methods

```python
# Shallow copy
a_copy = a.copy()
a_copy
```

```python
# True if no common element
a.isdisjoint(b)
```

```python
# a ⊆ b
a.issubset(b)
```

```python
# a ⊇ b
a.issuperset(b)
```

### 7. What Sets Cannot Do

- Cannot be indexed
- Cannot be sliced
- Cannot contain mutable elements (like lists/dicts)
- Cannot keep duplicates

#### Example:

```python
# becomes {1, 2, 3}
s = {1, 2, 2, 3}
s
```

### 8. When to Use Sets?

Use sets when you need:

- Fast membership testing
- Removing duplicates
- Mathematical operations
- Unique collection

### 9. Examples

```python
# Remove duplicates from a list:
nums = [1,2,2,3,3]
unique = list(set(nums))
unique
```

```python
# Check membership:
"apple" in {"apple", "banana"}
```

### 10. Set Comprehension

```python
s = {x*x for x in range(5)}
s
```
